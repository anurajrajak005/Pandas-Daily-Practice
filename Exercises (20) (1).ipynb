```python
top_items = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(10)
top_items
```

```python
top_items = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(10)

plt.bar(top_items.index, top_items.values)
plt.title("Top 10 Items by Quantity")
plt.xlabel("Item Name")
plt.ylabel("Quantity")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.scatter(df['quantity'], df['item_price'])
plt.title("Quantity vs Item Price")
plt.xlabel("Quantity")
plt.ylabel("Item Price")
plt.show()
```

```python
corr = df[['order_id', 'quantity', 'item_price']].corr()

plt.imshow(corr)
plt.title("Correlation Matrix")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.show()
```

```python
top5 = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(5)

plt.pie(top5.values, labels=top5.index, autopct='%1.1f%%')
plt.title("Top 5 Items Quantity Share")
plt.show()
```

```python
stats = df.groupby('quantity')['item_price'].agg(['mean', 'std']).head(10)

plt.errorbar(stats.index, stats['mean'], yerr=stats['std'], fmt='o')
plt.title("Mean Item Price with Error Bar")
plt.xlabel("Quantity")
plt.ylabel("Mean Price")
plt.show()
```

```python
plt.boxplot(df['item_price'])
plt.title("Boxplot of Item Price")
plt.ylabel("Item Price")
plt.show()
```
