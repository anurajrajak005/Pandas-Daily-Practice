```python
df = pd.read_csv("C:/Users/Asus/OneDrive/Apps/Desktop/adv-python/tips - tips.csv")
df
```

```python
# Display 
df.head()
# basis Plot 
# 1 . histogram of total bill (Univariate)
plt.figure()
plt.hist(df['total_bill'])
# plt.title('Distribution of total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
# Bar Chart : average tip by day 
avg_tip_day = df.groupby('day')['tip'].mean()
plt.figure()
avg_tip_day.plot(kind='bar')
# plt.title ('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

```python
#3. Scatter plot: total bili vs tip

plt.figure()
plt.scatter(df['total_bill'], df['tip'])
# plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
#4. Boxplot: total_bill by time

plt.figure()
df.boxplot(column='total_bill', by='time')
plt.title('Total Bill by Time')
plt.suptitle('')
plt.xlabel('Time')
plt.ylabel('Total Bill')
plt.show()
```

```python
 # how does average tip change accros days (as a trend)?
 # plt .plot it create the line plot
avg_tip_day = df.groupby('day')['tip'].mean()

plt.figure()
plt.plot(avg_tip_day,marker='o')
plt.title('Average Tip Trend by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

```python
#what percentage of customers are smokers vs non-smokers?
smoker_count = df['smoker'].value_counts()
plt.figure()
plt.pie(smoker_count.values, labels=smoker_count.index, autopct='%1.1f%%')
plt.title('Smoker vs Non-Smoker Distribution')
plt.show()
```
